```python
import numpy as np

class_a = np.array([78, 85, 92, 61, 74, 88, 55, 96, 70, 83])
class_b = np.array([65, 90, 72, 88, 45, 79, 93, 67, 84, 71])

print("Class A stats:")
print(f"  Mean:  {np.mean(class_a):.1f}")
print(f"  Max:   {np.max(class_a)}")
print(f"  Min:   {np.min(class_a)}")

print("Class B stats:")
print(f"  Mean:  {np.mean(class_b):.1f}")
print(f"  Max:   {np.max(class_b)}")
print(f"  Min:   {np.min(class_b)}")

combined = np.concatenate([class_a, class_b])
passing_a = class_a[class_a >= 60]
passing_b = class_b[class_b >= 60]
print(f"Class A: {len(passing_a)}/10 passed")
print(f"Class B: {len(passing_b)}/10 passed")
print(f"Combined mean: {np.mean(combined):.1f}")
print(f"Top score overall: {np.max(combined)}")

curved = np.minimum(combined + 5, 100)

a_count = int(np.sum(curved >= 90))
b_count = int(np.sum((curved >= 75) & (curved < 90)))
c_count = int(np.sum((curved >= 60) & (curved < 75)))
f_count = int(np.sum(curved < 60))

print("Grade distribution after +5 curve:")
print(f"  A (90-100): {a_count}")
print(f"  B (75-89):  {b_count}")
print(f"  C (60-74):  {c_count}")
print(f"  F (<60):    {f_count}")
print(f"New mean: {np.mean(curved):.1f}")
```
